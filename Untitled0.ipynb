```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

# Create a degree-2 polynomial regression model
mymodel = np.poly1d(np.polyfit(x, y, 2))

# Get polynomial coefficients
a, b, c = mymodel.coefficients

# Prediction function
def f(x_value):
    return mymodel(x_value)

# Display the polynomial equation
print(f"f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

# Predict speed at 5:00 PM, represented by x = 17
time_to_predict = 17
predicted_speed = f(time_to_predict)

# Print the predicted speed
print(f"Predicted speed at 17:00: {predicted_speed:.2f}")

# Create values for the regression curve
myline = np.linspace(1, 22, 100)

# Plot the original data
plt.figure(figsize=(10, 6))
plt.scatter(x, y, color="blue", label="Actual data")

# Plot the polynomial regression curve
plt.plot(
    myline,
    f(myline),
    color="red",
    label="Polynomial regression"
)

# Plot the predicted point at x = 17
plt.scatter(
    time_to_predict,
    predicted_speed,
    color="green",
    edgecolors="black",
    s=150,
    zorder=5,
    label=f"Prediction at 17:00 = {predicted_speed:.2f}"
)

# Display the predicted value on the graph
plt.annotate(
    f"{predicted_speed:.2f}",
    xy=(time_to_predict, predicted_speed),
    xytext=(10, 10),
    textcoords="offset points",
    fontsize=12,
    fontweight="bold",
    color="green",
    arrowprops=dict(
        arrowstyle="->",
        color="green"
    )
)

plt.title("Polynomial Regression: Vehicle Speed at 17:00")
plt.xlabel("Time")
plt.ylabel("Speed")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
