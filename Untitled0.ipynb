```python

import numpy as np
import matplotlib.pyplot as plt

x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

mymodel = np.poly1d(np.polyfit(x, y, 2))

a, b, c = mymodel.coefficients

def f(x_value):
    return mymodel(x_value)

print(f"f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

myline = np.linspace(1, 22, 100)

plt.scatter(x, y)
plt.plot(myline, f(myline), color='red')
plt.title("Polynomial Regression (degree=2)")
plt.show()
```

```python
# Time to predict: 5:00 PM = 17:00
time_to_predict = 17

# Predict the vehicle speed using the existing mymodel
predicted_speed = mymodel(time_to_predict)

# Print the predicted speed
print(f"Predicted vehicle speed at 17:00: {predicted_speed:.2f}")

# Plot the predicted point
plt.scatter(
    time_to_predict,
    predicted_speed,
    color="red",
    s=120,
    zorder=5,
    label=f"Predicted speed at 17:00 = {predicted_speed:.2f}"
)

# Display the value on the graph
plt.annotate(
    f"{predicted_speed:.2f}",
    xy=(time_to_predict, predicted_speed),
    xytext=(10, 10),
    textcoords="offset points",
    fontsize=12,
    color="red",
    arrowprops=dict(arrowstyle="->", color="red")
)

plt.legend()
plt.show()
```

```python
# ============================================================
# Vehicle speed prediction at 5:00 PM using Google Colab
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files
from sklearn.linear_model import LinearRegression

# ------------------------------------------------------------
# 1. Upload your CSV or Excel dataset
# ------------------------------------------------------------

uploaded = files.upload()
file_name = next(iter(uploaded))

if file_name.lower().endswith(".csv"):
    df = pd.read_csv(file_name)
elif file_name.lower().endswith((".xlsx", ".xls")):
    df = pd.read_excel(file_name)
else:
    raise ValueError("Please upload a CSV or Excel file.")

print("Dataset loaded successfully.")
print("Columns found:")
print(list(df.columns))
display(df.head())

# ------------------------------------------------------------
# 2. Automatically find the time and speed columns
# ------------------------------------------------------------

def find_column(columns, keywords):
    for column in columns:
        column_text = str(column).lower().strip()
        if any(keyword in column_text for keyword in keywords):
            return column
    return None

time_column = find_column(
    df.columns,
    ["time", "hour", "timestamp", "date"]
)

speed_column = find_column(
    df.columns,
    ["speed", "velocity", "kmph", "km_h", "km/h"]
)

if time_column is None or speed_column is None:
    print("\nThe automatic column detection did not find the required columns.")
    print("Available columns:", list(df.columns))
    print("\nSet these two variables manually, for example:")
    print("time_column = 'Time'")
    print("speed_column = 'Speed'")
    raise ValueError(
        "The dataset must contain one time column and one vehicle-speed column."
    )

print(f"\nTime column selected: {time_column}")
print(f"Speed column selected: {speed_column}")

# ------------------------------------------------------------
# 3. Convert time values into decimal hours
# ------------------------------------------------------------

def convert_time_to_decimal(value):
    if pd.isna(value):
        return np.nan

    # If the time is already numeric
    if isinstance(value, (int, float, np.integer, np.floating)):
        value = float(value)

        # Values from 0 to 24 are treated as decimal hours
        if 0 <= value <= 24:
            return value

        # Values such as 1020 are treated as minutes after midnight
        if 0 <= value <= 1440:
            return value / 60

    # Convert text such as 17:00, 5:00 PM, or 17:00:00
    text = str(value).strip()

    parsed = pd.to_datetime(text, errors="coerce")

    if pd.isna(parsed):
        return np.nan

    return parsed.hour + parsed.minute / 60 + parsed.second / 3600

df["time_decimal"] = df[time_column].apply(convert_time_to_decimal)
df["speed_numeric"] = pd.to_numeric(df[speed_column], errors="coerce")

# Remove invalid rows
data = df[["time_decimal", "speed_numeric"]].dropna().copy()

if len(data) < 2:
    raise ValueError(
        "At least two valid rows containing time and speed are required."
    )

# ------------------------------------------------------------
# 4. Train the linear regression model
# ------------------------------------------------------------

X = data[["time_decimal"]]
y = data["speed_numeric"]

model = LinearRegression()
model.fit(X, y)

# ------------------------------------------------------------
# 5. Predict speed at 5:00 PM
# ------------------------------------------------------------

time_to_predict = 17.0  # 17:00 = 5:00 PM

predicted_speed = model.predict(
    pd.DataFrame({"time_decimal": [time_to_predict]})
)[0]

print("\nPrediction result")
print("-----------------")
print(f"Predicted vehicle speed at 17:00: {predicted_speed:.2f}")

# ------------------------------------------------------------
# 6. Create the graph and display the prediction
# ------------------------------------------------------------

minimum_time = min(data["time_decimal"].min(), time_to_predict)
maximum_time = max(data["time_decimal"].max(), time_to_predict)

time_range = np.linspace(
    minimum_time,
    maximum_time,
    200
)

line_data = pd.DataFrame({"time_decimal": time_range})
predicted_line = model.predict(line_data)

plt.figure(figsize=(12, 7))

# Original observations
plt.scatter(
    data["time_decimal"],
    data["speed_numeric"],
    color="blue",
    s=60,
    label="Observed vehicle speed"
)

# Regression line
plt.plot(
    time_range,
    predicted_line,
    color="green",
    linewidth=2,
    label="Regression line"
)

# Prediction point at 17:00
plt.scatter(
    time_to_predict,
    predicted_speed,
    color="red",
    edgecolors="black",
    s=180,
    zorder=5,
    label=f"Prediction at 17:00: {predicted_speed:.2f}"
)

# Display value beside the red point
plt.annotate(
    f"{predicted_speed:.2f}",
    xy=(time_to_predict, predicted_speed),
    xytext=(12, 12),
    textcoords="offset points",
    fontsize=13,
    fontweight="bold",
    color="red",
    arrowprops={
        "arrowstyle": "->",
        "color": "red",
        "linewidth": 1.5
    }
)

# Format x-axis
plt.xlabel("Time")
plt.ylabel("Vehicle speed")
plt.title("Vehicle Speed Prediction at 5:00 PM")

# Show common time labels
tick_positions = np.arange(0, 25, 2)
tick_labels = [f"{int(hour):02d}:00" for hour in tick_positions]

plt.xticks(tick_positions, tick_labels)
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
