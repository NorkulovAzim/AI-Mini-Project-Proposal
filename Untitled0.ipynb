```python

import numpy as np
import matplotlib.pyplot as plt

x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

mymodel = np.poly1d(np.polyfit(x, y, 2))

a, b, c = mymodel.coefficients

def f(x_value):
    return mymodel(x_value)

print(f"f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

myline = np.linspace(1, 22, 100)

plt.scatter(x, y)
plt.plot(myline, f(myline), color='red')
plt.title("Polynomial Regression (degree=2)")
plt.show()
```

```python
# Time to predict: 5:00 PM = 17:00
time_to_predict = 17

# Predict the vehicle speed using the existing mymodel
predicted_speed = mymodel(time_to_predict)

# Print the predicted speed
print(f"Predicted vehicle speed at 17:00: {predicted_speed:.2f}")

# Plot the predicted point
plt.scatter(
    time_to_predict,
    predicted_speed,
    color="red",
    s=120,
    zorder=5,
    label=f"Predicted speed at 17:00 = {predicted_speed:.2f}"
)

# Display the value on the graph
plt.annotate(
    f"{predicted_speed:.2f}",
    xy=(time_to_predict, predicted_speed),
    xytext=(10, 10),
    textcoords="offset points",
    fontsize=12,
    color="red",
    arrowprops=dict(arrowstyle="->", color="red")
)

plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Define variables x and y from previous cells if df is not defined
try:
    df
except NameError:
    # Using existing data from your notebook variables x and y
    x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
    y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]
    df = pd.DataFrame({"Time": x, "Speed": y})

# View your data and column names
print(df.head())
print("Columns:", list(df.columns))

# Change these two names if your columns have different names
TIME_COLUMN = "Time"
SPEED_COLUMN = "Speed"

# Convert time values to decimal hours
def time_to_hours(value):
    if pd.isna(value):
        return np.nan

    # If time is already numeric, for example 8, 12.5, or 17
    if isinstance(value, (int, float, np.integer, np.floating)):
        return float(value)

    # Convert values such as "17:00", "5:00 PM", or "17:00:00"
    parsed_time = pd.to_datetime(str(value), errors="coerce")

    if pd.isna(parsed_time):
        return np.nan

    return (
        parsed_time.hour
        + parsed_time.minute / 60
        + parsed_time.second / 3600
    )

# Prepare the data
data = pd.DataFrame({
    "time_hours": df[TIME_COLUMN].apply(time_to_hours),
    "speed": pd.to_numeric(df[SPEED_COLUMN], errors="coerce")
}).dropna()

# Check that enough data exists
if len(data) < 2:
    raise ValueError("At least two valid time-speed rows are required.")

# Train the model
X = data[["time_hours"]]
y = data["speed"]

model = LinearRegression()
model.fit(X, y)

# Predict speed at 5:00 PM
time_to_predict = 17.0

prediction_data = pd.DataFrame({
    "time_hours": [time_to_predict]
})

predicted_speed = model.predict(prediction_data)[0]

# Print the predicted speed
print(f"Predicted vehicle speed at 17:00: {predicted_speed:.2f}")

# Create values for the regression line
time_values = np.linspace(
    min(data["time_hours"].min(), time_to_predict),
    max(data["time_hours"].max(), time_to_predict),
    200
)

line_data = pd.DataFrame({"time_hours": time_values})
speed_values = model.predict(line_data)

# Plot original data
plt.figure(figsize=(12, 7))

plt.scatter(
    data["time_hours"],
    data["speed"],
    color="blue",
    s=70,
    label="Actual data"
)

# Plot regression line
plt.plot(
    time_values,
    speed_values,
    color="green",
    linewidth=2,
    label="Regression line"
)

# Plot prediction at 17:00
plt.scatter(
    time_to_predict,
    predicted_speed,
    color="red",
    edgecolor="black",
    s=180,
    zorder=5,
    label=f"Predicted speed at 17:00 = {predicted_speed:.2f}"
)

# Display predicted value on graph
plt.annotate(
    f"{predicted_speed:.2f}",
    xy=(time_to_predict, predicted_speed),
    xytext=(12, 12),
    textcoords="offset points",
    fontsize=13,
    fontweight="bold",
    color="red",
    arrowprops=dict(
        arrowstyle="->",
        color="red"
    )
)

plt.xlabel("Time")
plt.ylabel("Vehicle speed")
plt.title("Vehicle Speed Prediction at 5:00 PM")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
